```python
df['Rating'].unique() #Valores únicos dentro de esa columna
```

```python
np.sort(df['Rating'].unique()) # 'sort' ordena de menor a mayor
```

```python
df[df.Rating>8].tail() # Obtiene los valores mayores a 8 de la columna Rating (el tail sirve para los últimos, el head son los primeros 5)
```

```python
df.Rating.value_counts() #Value_counts sirve para ver la cantidad de veces que se repite
```

```python
counts = df.Rating.value_counts()

list(counts.index)
```

```python
counts.values #Muestra la cantidad de veces  de un valor en cierta columna
```

```python
counts_df = pd.DataFrame({'calif':counts.index, 'frq':counts.values}) #Prsentacion de datos en tabla

counts_df.head()
```

```python
import seaborn as sns # Seaborn genera gráficos estadísticos, con una mejor UX
import matplotlib.pyplot as plt # Sirve para personalizar la gráfica (ambas se complementan)
```

```python
sns.barplot(
    data = counts_df,
    x = 'calif',
    y = 'frq'
)

plt.title('Frecuencia de calificaciones')

plt.show()
```

```python
plt.figure(figsize=(10, 6)) # le da tamaños a la grafica
sns.histplot(
    data=df, 
    x='Rating', 
    kde=True, # le agrega una curva de densidad sobre las mismas graficas
    bins=25, 
    color = 'skyblue'
)
plt.title('Distribución de calificaciones de películas') # titulo
plt.xlabel('Califiación') # Título para el eje x
plt.ylabel('Cantidad de películas') # Titulo para el eje y
plt.grid(axis='y', linestyle='--', alpha=0.5)
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(data=df, x='Rating', y='Revenue (Millions)', alpha=0.6, color = 'coral', edgecolor = 'black')

plt.title('Relación entre calificaciones e ingresos')
plt.xlabel('Calificacion')
plt.ylabel('Ingresos')
plt.grid(True, linestyle='--', alpha = 1)
```

```python
plt.figure(figsize=(10,6))
sns.kdeplot(data=df, x='Runtime (Minutes)', fill=True, color='darkviolet', alpha = 0.5)

plt.title('Distribución Suavizada de la duración de Películas')
plt.xlabel('Duración en minutos')
plt.ylabel('Densidad de probabilidad')
```
